# 1 prices.condition Cleaning

```python
df_clean["condition_clean"] = (
    df_clean["prices.condition"]
    .str.strip()
    .str.lower()
)
df_clean["condition_clean"].value_counts(dropna=False)
def clean_condition(value):
    if pd.isna(value):
        return "unknown"
    
    if "refurbished" in value:
        return "refurbished"
    
    if value == "pre-owned" or value == "used":
        return "used"
    
    if value.startswith("new"):
        return "new"
    
    return "other"

df_clean["condition_clean"] = df_clean["condition_clean"].apply(clean_condition)
df_clean["condition_clean"].value_counts()
```

# 2 prices.availability cleaning

```python
df_clean["availability_clean"] = (
    df_clean["prices.availability"]
    .str.strip()
    .str.lower()
)
df_clean["availability_clean"].value_counts(dropna=False)

def clean_availability(value):
    if value in ["in stock", "yes", "true", "32 available", "7 available"]:
        return "available"
    
    elif value in ["out of stock", "no", "false", "sold", "retired"]:
        return "unavailable"
    
    elif value in ["special order", "more on the way"]:
        return "limited/pending"
    
    else:
        return "unknown"

df_clean["availability_clean"] = (
    df_clean["availability_clean"].apply(clean_availability)
)
df_clean["availability_clean"].value_counts()
```

# 3 Brand cleaning

```python
df_clean["brand_clean"] = (
    df_clean["brand"]
    .str.strip()
    .str.title()
)
df_clean["brand_clean"] = df_clean["brand_clean"].replace(
    {"Wd": "Western Digital"}
)
print("Original unique brands:", df_clean["brand"].nunique())
print("Clean unique brands:", df_clean["brand_clean"].nunique())
```

# 4 Merchant cleaning

```python
df_clean["merchant_clean"] = (
    df_clean["prices.merchant"]
    .str.strip()
    .str.lower()
)
print("Original unique merchants:", df_clean["prices.merchant"].nunique())
print("Clean unique merchants:", df_clean["merchant_clean"].nunique())
```

# 5 Shipping cleaning

```python
df_clean["shipping_clean"] = (
    df_clean["prices.shipping"]
    .str.strip()
    .str.lower()
)
df_clean["shipping_clean"].value_counts(dropna=False).head(20)

#category defining 
def categorize_shipping(value):
    if pd.isna(value):
        return "unknown"
    
    if "free" in value:
        if "over" in value or "orders 35" in value or "orders $35" in value:
            return "conditional free"
        else:
            return "free"
    
    return "paid/other"

df_clean["shipping_type"] = (
    df_clean["shipping_clean"].apply(categorize_shipping)
)
df_clean["shipping_type"].value_counts()
```

```python
products = (
    df_clean[
        ["id", "name", "brand_clean", "categories"]
    ]
    .drop_duplicates(subset="id")
    .copy()
)

print("Product rows:", len(products))
products.head()

products["name"].head(30).tolist()
```

```python
print("Unique products:", len(products))
print("Unique category combinations:", products["categories"].nunique())
```

```python
keywords = [
    "computer", "laptop", "tablet",
    "tv", "television",
    "audio", "speaker", "headphone",
    "camera",
    "mobile", "phone",
    "car", "auto"
]

for word in keywords:
    count = products["categories"].str.contains(
        word, case=False, na=False
    ).sum()
    print(f"{word}: {count}")


    
```

```python
def assign_category(row):
    text = (
        str(row["name"]) + " " + str(row["categories"])
    ).lower()

    if any(word in text for word in [
        "car electronics", "car audio", "auto electronics",
        "automotive", "car speaker"
    ]):
        return "Automotive Electronics"

    elif any(word in text for word in [
        "camera", "camcorder", "photography"
    ]):
        return "Cameras & Photography"

    elif any(word in text for word in [
        "phone", "mobile", "smartphone", "iphone"
    ]):
        return "Mobile & Phones"

    elif any(word in text for word in [
        "computer", "laptop", "tablet", "desktop",
        "ssd", "hard drive", "memory"
    ]):
        return "Computers & Tablets"

    elif any(word in text for word in [
        "tv", "television", "home theater", "projector"
    ]):
        return "TV & Home Theater"

    elif any(word in text for word in [
        "audio", "speaker", "headphone", "earbud", "stereo"
    ]):
        return "Audio"

    else:
        return "Other Electronics"


products["category_clean"] = products.apply(assign_category, axis=1)
products["category_clean"].value_counts()
```

```python
products[
    products["category_clean"] == "Other Electronics"
][["name", "brand_clean", "categories"]]
```

```python
df_clean = df_clean.merge(
    products[["id", "category_clean"]],
    on="id",
    how="left"
)
print("Rows:", len(df_clean))

print("Missing categories:", df_clean["category_clean"].isna().sum())
df_clean["category_clean"].value_counts()
```

## Data Cleaning Summary

The raw electronics pricing dataset was assessed and cleaned before analysis.

### Cleaning Steps
- Removed columns with extremely high missing-value rates that were not useful for the analysis.
- Checked for exact duplicate records; no exact duplicates were found.
- Standardized product condition into broader groups: New, Used, Refurbished, and Other.
- Standardized product availability into Available, Unavailable, Limited/Pending, and Unknown.
- Standardized brand and merchant text to reduce inconsistencies caused by capitalization and whitespace.
- Grouped shipping information into Free, Conditional Free, Paid/Other, and Unknown.
- Restricted the analysis to USD listings. One CAD listing was excluded rather than converted because it represented only one record.
- Parsed `prices.dateSeen` and selected the most recent timestamp when multiple observation dates were stored in a single record.
- Created a representative `price` variable using the midpoint of `prices.amountMin` and `prices.amountMax`.
- Investigated unusually low prices and removed 13 listings below $5 after confirming that they represented implausible prices for major electronics products.
- Investigated high-priced observations and retained them because they represented plausible premium electronics such as televisions and laptops.
- Created broader business-friendly product categories using product names and existing category information.
- Preserved repeated product listings because the same product can appear across different merchants, prices, and observations.

### Final Dataset
- Listings: 7,235
- Unique products: 835
- Core analysis columns with missing values: 0
- Exact duplicate rows: 0
- Currency: USD
- Price range: $6.45–$6,498.99

The cleaned dataset was saved as `data/processed/electronics_pricing_clean.csv` for exploratory data analysis.
